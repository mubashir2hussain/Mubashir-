```
#4
def get_money(currency: str = "Euro", amount: int | float = 500) -> str:
    """
    Returns a formatted string representing an amount of money.

    Parameters:
        currency (str): The name of the currency (default: "Euro").
        amount (int | float): The amount of money (default: 500).
    """
    amount_str = str(amount)

    # Handle pluralization (e.g., 1 Euro vs 500 Euros)
    if amount_str != "1" and not currency.endswith("s"):
        currency += "s"

    return f"Ok, here are {amount} {currency}"


# --- Examples ---

# 1. Using both defaults (Euro, 500)
print(get_money())
# Output: Ok, here are 500 Euros

# 2. Providing a custom currency (uses default amount: 500)
print(get_money("Dollar"))
# Output: Ok, here are 500 Dollars

# 3. Providing both custom currency and custom amount
print(get_money("Yen", 1000))
# Output: Ok, here are 1000 Yens

# 4. Testing singular amount (N = 1)
print(get_money("Euro", 1))
# Output: Ok, here are 1 Euro
```
